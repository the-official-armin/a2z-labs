**Matrix Multiplication**

2 types of multiplication
1. Element wise multiplication
2. Matrix multiplication (Dot Product)

2 rules for matmul
1. **Inner dimensions** must match
- `(3, 2) @ (3, 2)` wont work ( Below code )
- `(2, 3) @ (3, 2)` will work

2. The resulting matrix has a shape of the **outer dimensions**:
- `(2, 3) @ (3, 2)` -> `(2,2)`



```python
torch.matmul(torch.randn(3, 2), torch.randn(3, 2))
```

```python
# Element wise multiplication

print(tensor, '*', tensor)
print(f'Equals: {tensor*tensor}')
```

> **Key tip**: Most common errors in DL: shape errors!

```python
# Shapes for matrix multiplication

tensor_A = torch.tensor([[1,2],
                         [3,4],
                         [5,6]])

tensor_B = torch.tensor([[7,10],
                         [8,11],
                         [9,12]])

torch.matmul(tensor_A, tensor_B)
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor shape issues we can manipulate the shape of our tensors using a **transpose**.

```python
tensor_B.T, tensor_B
```

```python
tensor_B.T.shape
```

```python
torch.matmul(tensor_A, tensor_B.T)
```

### Reshaping, stacking, squeezing, and unsqueezing

- Reshape: reshapes an input tensor to a defined shape
- View: Return a view of an input tensor of a certain shape but keep the same memory as a original tensor
- Stacking: combine multiple tensors on top of each other or side by side
- squeeze: remove all `1` dimensions from a tensor
- Unsqueeze: add a `1` dimension to a target tensor
- Permute: return a view of the input w/ dimensions permuted in a certain way.

```python
x = torch.arange(1., 11.)
x, x.shape
```

```python
x_reshaped = x.reshape(1, 10)
x_reshaped, x_reshaped.shape
```

```python
x_reshaped = x.reshape(10, 1)
x_reshaped, x_reshaped.shape
```

```python
x_reshaped = x.reshape(1, 10)
x_reshaped, x_reshaped.shape
```

```python
z = x.view(1, 10)
z, z.shape
```

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=1)
x_stacked
```

```python
# Squeeze
squeezed = x_reshaped.squeeze()

squeezed, squeezed.shape
```

```python
# Unsqueeze
unsqueezed = squeezed.unsqueeze(dim=0)
unsqueezed, unsqueezed.shape
```

```python
# Permute

x_og = torch.randn(size=(224, 224, 3))

x_permuted = x_og.permute(2, 0, 1)
x_og.shape, x_permuted.shape
```

### Pytorch tensors and Numpy

Numpy is a popular scientific python numerical computing library

Because of this, pytorch has functionality to interact with it.

- Data in Numpy, want in pytorch tensor `torch.from_numpy(ndarray)`
- Pytorch tensor -> Numpy -> torch.Tensor.numpy()

```python
# Numpy array to tensor
import torch
import numpy as np

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
array.dtype
```

```python
torch.arange(1.0, 8.0).dtype
```

```python
array = array + 1
array, tensor
```

```python
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

## Reproducebility

To reduce randomness in NN and pytorch comes the concept of **random seed**

Random seed "flavors" the randomness.


```python
random_A = torch.rand(3,4)
random_B = torch.rand(3, 4)

print(random_A)
print(random_B)
print(random_A == random_B)
```

```python
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_C = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_D = torch.rand(3,4)

print(random_C)
print(random_D)
print(random_C == random_D)
```
